# Basket of Words trained model

```python
import sys
import os
import matplotlib.pyplot as plt

os.chdir(os.path.abspath(os.path.join(os.getcwd(), "..")))


from src.data.processor import unzip_data_extract_contents
from src.model.train_model import train_bow_logreg
from src.model.predict_model import predict_sentiment_bow

test_archive, train_archive, unsup_archive, imdb_vocab, imdb_expected_rating = unzip_data_extract_contents()
train_bow_logreg(test_archive, train_archive, imdb_vocab)

sample_text = "bad movie"
print("BoW Prediction:", predict_sentiment_bow(sample_text, imdb_vocab))

sample_text = "great ever"
print(f"BoW Prediction: {sample_text}", predict_sentiment_bow(sample_text, imdb_vocab))
```

### Prediction model with Hyperparameter tuning

```python

sample_text = "bad movie"
print("BoW Prediction:", predict_sentiment_bow(sample_text, imdb_vocab, with_params=True))

sample_text = "great ever"
print(f"BoW Prediction: {sample_text}", predict_sentiment_bow(sample_text, imdb_vocab, with_params=True))
```

```python
y_train_true = [r['type'] for r in train_archive.reviews]
y_train_pred = [predict_sentiment_bow(r['contents'], imdb_vocab) for r in train_archive.reviews]
train_accuracy = sum([yt == yp for yt, yp in zip(y_train_true, y_train_pred)]) / len(y_train_true)

# Testing set accuracy
y_test_true = [r['type'] for r in test_archive.reviews]
y_test_pred = [predict_sentiment_bow(r['contents'], imdb_vocab) for r in test_archive.reviews]
test_accuracy = sum([yt == yp for yt, yp in zip(y_test_true, y_test_pred)]) / len(y_test_true)

import matplotlib.pyplot as plt

plt.bar(['Train', 'Test'], [train_accuracy, test_accuracy], color=['skyblue', 'orange'])
plt.ylim(0, 1)
plt.ylabel('Accuracy')
plt.title('BoW Model Accuracy: Train vs Test')
plt.show()
```
